```python
# =======================================================
# 2. CARREGAR E FUNDIR AS PLANILHAS
# =======================================================
def carregar_planilhas(caminhos):
    dfs = []
    for p in caminhos:
        try:
            colunas_esperadas = ['ANIMAL', 'L_pelo', 'a_pelo', 'b_pelo', 'L_pele', 'a_pele', 'b_pele']
            try:
                # Tenta ler como o formato novo (Julho 2026, utf-8, já sem o ID extra)
                temp = pd.read_csv(p, encoding='utf-8')
                temp.columns = colunas_esperadas
                dfs.append(temp)
            except Exception:
                # Cai aqui para as antigas de Dezembro/Fevereiro, forçando exatamente 8 colunas no names
                col_names_antigas = ['ID_Seq', 'ANIMAL', 'L_pelo', 'a_pelo', 'b_pelo', 'L_pele', 'a_pele', 'b_pele']
                temp = pd.read_csv(p, encoding='latin-1', names=col_names_antigas, skiprows=2, on_bad_lines='skip')
                temp = temp.drop(columns=['ID_Seq'])
                dfs.append(temp)
        except Exception as e:
            print(f"Erro inesperado ao carregar {p}: {e}")
    if not dfs: return pd.DataFrame()
    df_final = pd.concat(dfs, ignore_index=True)
    df_final = df_final.dropna(subset=['ANIMAL'])
    df_final['ANIMAL'] = df_final['ANIMAL'].astype(float).astype(int).astype(str)
    cols_lab = ['L_pelo', 'a_pelo', 'b_pelo', 'L_pele', 'a_pele', 'b_pele']
    for c in cols_lab:
        df_final[c] = pd.to_numeric(df_final[c], errors='coerce')
    return df_final
```

```python
# =======================================================
# 3. EXTRAÇÃO MULTICANAL (RGB + HSV + LAB) COM GRID
# =======================================================
def aplicar_white_balance(img_bgr):
    """
    Aplica o algoritmo Gray World para corrigir a iluminação da imagem,
    removendo tons amarelados ou azulados causados pela luz ambiente.
    """
    resultado = np.copy(img_bgr).astype(np.float32)

    # Calcula a média de cada canal de cor (Azul, Verde e Vermelho)
    avg_b = np.mean(resultado[:,:,0])
    avg_g = np.mean(resultado[:,:,1])
    avg_r = np.mean(resultado[:,:,2])

    # Média global de todos os canais
    avg_global = (avg_b + avg_g + avg_r) / 3.0

    # Evita divisão por zero caso a imagem seja um buraco negro rs
    if avg_b == 0 or avg_g == 0 or avg_r == 0:
        return img_bgr

    # Multiplica cada canal por um fator de correção
    resultado[:,:,0] = np.clip(resultado[:,:,0] * (avg_global / avg_b), 0, 255)
    resultado[:,:,1] = np.clip(resultado[:,:,1] * (avg_global / avg_g), 0, 255)
    resultado[:,:,2] = np.clip(resultado[:,:,2] * (avg_global / avg_r), 0, 255)

    return resultado.astype(np.uint8)


def extrair_grid_multicanal(caminho_img, num_partes=16):
    img_bgr = cv2.imread(caminho_img)
    if img_bgr is None: return []

    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    h_orig, w_orig = gray.shape

    mascara = np.zeros_like(gray)
    cv2.circle(mascara, (w_orig//2, h_orig//2), int(min(h_orig, w_orig)*0.28), 255, -1)

    gray_foco = gray.copy()
    gray_foco[mascara == 0] = 255
    blur = cv2.GaussianBlur(gray_foco, (51, 51), 0)

    thresh_val = 80 if "pele" in caminho_img.lower() else 165
    _, thresh = cv2.threshold(blur, thresh_val, 255, cv2.THRESH_BINARY_INV)

    contornos, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    if not contornos: return []

    maior_contorno = max(contornos, key=cv2.contourArea)
    x, y, w, h = cv2.boundingRect(maior_contorno)

    mx, my = int(w*0.1), int(h*0.1)
    y1, y2 = y + my, y + h - my
    x1, x2 = x + mx, x + w - mx

    if y2 <= y1 or x2 <= x1: return []

    crop = img_bgr[y1:y2, x1:x2]
    crop_corrigido = aplicar_white_balance(crop)

    crop_rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)

    # Criar versões nos outros espaços de cor
    crop_hsv = cv2.cvtColor(crop_rgb, cv2.COLOR_RGB2HSV)
    crop_lab_cv2 = cv2.cvtColor(crop_rgb, cv2.COLOR_RGB2LAB)

    grid = int(np.sqrt(num_partes))
    h_c, w_c = crop_rgb.shape[:2]
    step_h, step_w = h_c // grid, w_c // grid

    lista_features = []

    for i in range(grid):
        for j in range(grid):
            y_inicio = i * step_h
            y_fim = (i + 1) * step_h if i < grid - 1 else h_c
            x_inicio = j * step_w
            x_fim = (j + 1) * step_w if j < grid - 1 else w_c

            bloco_rgb = crop_rgb[y_inicio:y_fim, x_inicio:x_fim]
            if bloco_rgb.size == 0: continue

            bloco_hsv = crop_hsv[y_inicio:y_fim, x_inicio:x_fim]
            bloco_lab = crop_lab_cv2[y_inicio:y_fim, x_inicio:x_fim]

            rgb_m = np.mean(bloco_rgb, axis=(0,1))
            hsv_m = np.mean(bloco_hsv, axis=(0,1))
            lab_m = np.mean(bloco_lab, axis=(0,1))

            lista_features.append({
                'R': rgb_m[0], 'G': rgb_m[1], 'B': rgb_m[2],
                'H': hsv_m[0], 'S': hsv_m[1], 'V': hsv_m[2],
                'L_cv2': lab_m[0], 'a_cv2': lab_m[1], 'b_cv2': lab_m[2]
            })

    return lista_features
```

```python
# =======================================================
# 3. CROP E EXTRAÇÃO DA COR DA MANCHA
# =======================================================
def extrair_cor_pura_multicanal(caminho_img):
    img_bgr = cv2.imread(caminho_img)
    if img_bgr is None: return []

    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    max_h, max_w = gray.shape

    # 1. Isolar o círculo
    blur_luz = cv2.GaussianBlur(gray, (61, 61), 0)
    _, thresh_luz = cv2.threshold(blur_luz, 100, 255, cv2.THRESH_BINARY)

    kernel_luz = np.ones((45, 45), np.uint8)
    thresh_luz = cv2.morphologyEx(thresh_luz, cv2.MORPH_OPEN, kernel_luz)

    contornos_luz, _ = cv2.findContours(thresh_luz, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    if not contornos_luz: return []
    maior_contorno_luz = max(contornos_luz, key=cv2.contourArea)

    mascara_amostra = np.zeros_like(gray)
    cv2.drawContours(mascara_amostra, [maior_contorno_luz], -1, 255, thickness=cv2.FILLED)

    # 2. Pintar o cone de branco
    gray_isolado = gray.copy()
    gray_isolado[mascara_amostra == 0] = 255

    # 3. Encontrar a mancha escura (Threshold dinâmico pele/pelo)
    thresh_val = 80 if "pele" in caminho_img.lower() else 165

    blur_mancha = cv2.GaussianBlur(gray_isolado, (151, 151), 0)
    _, mascara_mancha = cv2.threshold(blur_mancha, thresh_val, 255, cv2.THRESH_BINARY_INV)

    kernel_mancha = np.ones((61, 61), np.uint8)
    mascara_mancha = cv2.morphologyEx(mascara_mancha, cv2.MORPH_CLOSE, kernel_mancha)
    mascara_mancha = cv2.morphologyEx(mascara_mancha, cv2.MORPH_OPEN, kernel_mancha)

    contornos_mancha, _ = cv2.findContours(mascara_mancha, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    if not contornos_mancha: return []
    maior_contorno_mancha = max(contornos_mancha, key=cv2.contourArea)

    # Coordenadas da mancha
    X_exato, Y_exato, W_exato, H_exato = cv2.boundingRect(maior_contorno_mancha)

    # 4. Adicionando Borda
    percentual_borda = 0.05
    pad_w = int(W_exato * percentual_borda)
    pad_h = int(H_exato * percentual_borda)

    y1 = max(0, Y_exato - pad_h)
    y2 = min(max_h, Y_exato + H_exato + pad_h)
    x1 = max(0, X_exato - pad_w)
    x2 = min(max_w, X_exato + W_exato + pad_w)

    crop_final = img_bgr[y1:y2, x1:x2]
    if crop_final.size == 0: return []

    crop_rgb = cv2.cvtColor(crop_final, cv2.COLOR_BGR2RGB)
    crop_hsv = cv2.cvtColor(crop_rgb, cv2.COLOR_RGB2HSV)
    crop_lab_cv2 = cv2.cvtColor(crop_rgb, cv2.COLOR_RGB2LAB)

    # =======================================================
    # MEDIANA E DESVIO PADRÃO
    # =======================================================
    # Usando axis=(0, 1) para calcular a mediana de todos os pixels na Altura e Largura
    rgb_m = np.median(crop_rgb, axis=(0, 1))
    hsv_m = np.median(crop_hsv, axis=(0, 1))
    lab_m = np.median(crop_lab_cv2, axis=(0, 1))

    rgb_s = np.std(crop_rgb, axis=(0, 1))
    hsv_s = np.std(crop_hsv, axis=(0, 1))
    lab_s = np.std(crop_lab_cv2, axis=(0, 1))

    return [{
        'R': rgb_m[0], 'G': rgb_m[1], 'B': rgb_m[2],
        'H': hsv_m[0], 'S': hsv_m[1], 'V': hsv_m[2],
        'L_cv2': lab_m[0], 'a_cv2': lab_m[1], 'b_cv2': lab_m[2],
        'R_std': rgb_s[0], 'G_std': rgb_s[1], 'B_std': rgb_s[2],
        'H_std': hsv_s[0], 'S_std': hsv_s[1], 'V_std': hsv_s[2],
        'L_std': lab_s[0], 'a_std': lab_s[1], 'b_std': lab_s[2]
    }]
```

```python
# =======================================================
# 4. MONTAR O DATASET
# =======================================================
print("1. Lendo as planilhas do colorímetro...")
df_colorimetro = carregar_planilhas(caminhos_planilhas)
print(f" -> {len(df_colorimetro)} registros carregados com sucesso.")

dataset_final = []
arquivos_sucesso = 0
erro_nome = 0
erro_planilha = 0
erro_opencv = 0
erro_lab_vazio = 0

todas_as_fotos = []
for pasta_atual in caminhos_fotos:
    if not os.path.exists(pasta_atual): continue
    for arquivo in os.listdir(pasta_atual):
        if arquivo.lower().endswith(('.jpg', '.png')):
            todas_as_fotos.append((pasta_atual, arquivo))

print(f"\n2. Processando {len(todas_as_fotos)} imagens...")

for pasta_atual, arquivo in tqdm(todas_as_fotos, desc="Extraindo Cores", unit="foto"):

    caminho_completo = os.path.join(pasta_atual, arquivo)

    match = re.search(r'\d+', arquivo)
    if match:
        animal_id = match.group()
    else:
        erro_nome += 1
        continue

    tipo = "pele" if "pele" in arquivo.lower() else "pelo"

    linha = df_colorimetro[df_colorimetro['ANIMAL'] == animal_id]
    if linha.empty:
        erro_planilha += 1
        continue

    if tipo == "pele":
        l, a, b = linha[['L_pele', 'a_pele', 'b_pele']].iloc[0]
    else:
        l, a, b = linha[['L_pelo', 'a_pelo', 'b_pelo']].iloc[0]

    if pd.isna(l):
        erro_lab_vazio += 1
        continue

    #features_lista = extrair_cor_pura_multicanal(caminho_completo)
    features_lista = extrair_grid_multicanal(caminho_completo, num_partes=9)
    if not features_lista:
        erro_opencv += 1
        continue

    for f in features_lista:
        dataset_final.append({
            'ANIMAL': animal_id,
            'R': f['R'], 'G': f['G'], 'B': f['B'],
            'H': f['H'], 'S': f['S'], 'V': f['V'],
            'L_cv2': f['L_cv2'], 'a_cv2': f['a_cv2'], 'b_cv2': f['b_cv2'],
            'L': l, 'a': a, 'b': b,
            'Tipo_Pele': 1 if tipo == "pele" else 0
        })
    arquivos_sucesso += 1

df_ml = pd.DataFrame(dataset_final)

print("\n\n=======================================================")
print("             DIAGNÓSTICO DE PERDAS                     ")
print("=======================================================")
print(f"SUCESSO: {arquivos_sucesso} fotos convertidas")
print(f"Falha ao entender o nome do arquivo: {erro_nome}")
print(f"Animal não encontrado nas planilhas: {erro_planilha}")
print(f"Célula vazia (sem valor LAB) no Excel: {erro_lab_vazio}")
print(f"OpenCV falhou ao achar a mancha da pele/pelo: {erro_opencv}")
print("=======================================================")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
import pandas as pd

print("\nAplicando Engenharia de Features e Treinando Especialistas...")
df_ia = df_ml.dropna().copy()

# ==========================================
# ESTRATÉGIA 3: CRIANDO NOVAS FEATURES MATEMÁTICAS
# ==========================================
# 1. Proporção Vermelho/Verde (Ajuda muito o modelo a achar o eixo 'a' do LAB)
# Somamos 0.000001 para evitar o erro de dividir por zero caso o G seja absoluto 0
df_ia['Razao_RG'] = df_ia['R'] / (df_ia['G'] + 0.000001)

# 2. Proporção Vermelho/Azul e Verde/Azul (Ajudam no eixo 'b' do LAB)
df_ia['Razao_RB'] = df_ia['R'] / (df_ia['B'] + 0.000001)
df_ia['Razao_GB'] = df_ia['G'] / (df_ia['B'] + 0.000001)

# 3. Diferença de interpretação de Luminosidade
df_ia['Dif_L_V'] = df_ia['L_cv2'] - df_ia['V']

# 4. Intensidade Total de Luz da Câmera
df_ia['Soma_RGB'] = df_ia['R'] + df_ia['G'] + df_ia['B']

# Atualizamos a lista de features (O que a IA vai ler)
features = [
    'R', 'G', 'B', 'H', 'S', 'V', 'L_cv2', 'a_cv2', 'b_cv2',
    'Razao_RG', 'Razao_RB', 'Razao_GB', 'Dif_L_V', 'Soma_RGB'
]
targets = ['L', 'a', 'b']

# ==========================================
# MODELO 1: ESPECIALISTA EM PELE
# ==========================================
df_pele = df_ia[df_ia['Tipo_Pele'] == 1]
X_pele = df_pele[features]
Y_pele = df_pele[targets]

X_train_pele, X_test_pele, Y_train_pele, Y_test_pele = train_test_split(X_pele, Y_pele, test_size=0.2, random_state=42)

rf_pele = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
rf_pele.fit(X_train_pele, Y_train_pele)
df_prev_pele = pd.DataFrame(rf_pele.predict(X_test_pele), columns=['L_pred', 'a_pred', 'b_pred'])

# ==========================================
# MODELO 2: ESPECIALISTA EM PELO
# ==========================================
df_pelo = df_ia[df_ia['Tipo_Pele'] == 0]
X_pelo = df_pelo[features]
Y_pelo = df_pelo[targets]

X_train_pelo, X_test_pelo, Y_train_pelo, Y_test_pelo = train_test_split(X_pelo, Y_pelo, test_size=0.2, random_state=42)

rf_pelo = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
rf_pelo.fit(X_train_pelo, Y_train_pelo)
df_prev_pelo = pd.DataFrame(rf_pelo.predict(X_test_pelo), columns=['L_pred', 'a_pred', 'b_pred'])

# ==========================================
# AVALIAÇÃO E RESULTADOS
# ==========================================
# Aqui está a correção usando o pd.concat para juntar os DataFrames perfeitamente!
Y_test_geral = pd.concat([Y_test_pele, Y_test_pelo])
prev_geral = pd.concat([df_prev_pele, df_prev_pelo])

print("\n================ RESULTADOS DOS ESPECIALISTAS ================")
print(f"Erro em L: ± {mean_absolute_error(Y_test_geral['L'], prev_geral['L_pred']):.2f}")
print(f"Erro em a: ± {mean_absolute_error(Y_test_geral['a'], prev_geral['a_pred']):.2f}")
print(f"Erro em b: ± {mean_absolute_error(Y_test_geral['b'], prev_geral['b_pred']):.2f}")
print(f"R² Score Geral: {r2_score(Y_test_geral, prev_geral):.4f}")
print("==============================================================")

# Imprimir a importância das features para o Pelo (que é o mais difícil)
#importancias_pelo = pd.DataFrame({'Canal': features, 'importancia': rf_pelo.feature_importances_}).sort_values(by='Importancia', ascending=False)
#print("\nFeatures mais importantes para o modelo:")
#print(importancias_pelo.head(5).to_string(index=False))
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
from scipy.stats import pearsonr

def calcular_delta_e_unitario(y_true, y_pred):
    diff = y_true.values - y_pred.values
    return np.sqrt(np.sum(diff**2, axis=1)).mean()

def bland_altman_stats(y_true, y_pred):
    diff = y_true - y_pred
    bias = np.mean(diff)
    std_diff = np.std(diff, ddof=1)
    loa = (bias - 1.96 * std_diff, bias + 1.96 * std_diff)
    return bias, loa

def imprimir_relatorio_avancado(y_true, y_pred):
    # MAE
    mae_L = mean_absolute_error(y_true['L'], y_pred['L_pred'])
    mae_a = mean_absolute_error(y_true['a'], y_pred['a_pred'])
    mae_b = mean_absolute_error(y_true['b'], y_pred['b_pred'])

    # R2
    r2_L = r2_score(y_true['L'], y_pred['L_pred'])
    r2_a = r2_score(y_true['a'], y_pred['a_pred'])
    r2_b = r2_score(y_true['b'], y_pred['b_pred'])

    # RMSE
    rmse_L = np.sqrt(mean_squared_error(y_true['L'], y_pred['L_pred']))
    rmse_a = np.sqrt(mean_squared_error(y_true['a'], y_pred['a_pred']))
    rmse_b = np.sqrt(mean_squared_error(y_true['b'], y_pred['b_pred']))

    # Pearson
    pearson_L, _ = pearsonr(y_true['L'], y_pred['L_pred'])
    pearson_a, _ = pearsonr(y_true['a'], y_pred['a_pred'])
    pearson_b, _ = pearsonr(y_true['b'], y_pred['b_pred'])

    # Delta E
    delta_e_mean = calcular_delta_e_unitario(y_true, y_pred)

    # Bland-Altman
    bias_L, loa_L = bland_altman_stats(y_true['L'], y_pred['L_pred'])
    bias_a, loa_a = bland_altman_stats(y_true['a'], y_pred['a_pred'])
    bias_b, loa_b = bland_altman_stats(y_true['b'], y_pred['b_pred'])

    print("\n=======================================================")
    print("             MÉTRICAS DE AVALIAÇÃO (CONJUNTO DE TESTE) ")
    print("=======================================================")
    print(f"Erro Médio Absoluto (MAE): L*={mae_L:.3f} | a*={mae_a:.3f} | b*={mae_b:.3f}")
    print(f"Coeficiente de Determinação (R2): L*={r2_L:.3f} | a*={r2_a:.3f} | b*={r2_b:.3f}")
    print(f"Raiz do Erro Quadrático Médio (RMSE): L*={rmse_L:.3f} | a*={rmse_a:.3f} | b*={rmse_b:.3f}")
    print(f"Coeficiente de Correlação (r): L*={pearson_L:.3f} | a*={pearson_a:.3f} | b*={pearson_b:.3f}")
    print(f"Diferença de Cor (Delta E) Médio: {delta_e_mean:.3f}")
    print("\n--- ANÁLISE DE BLAND-ALTMAN ---")
    print(f"L* -> Viés: {bias_L:.3f} | Limites de Concordância 95%: ({loa_L[0]:.3f}, {loa_L[1]:.3f})")
    print(f"a* -> Viés: {bias_a:.3f} | Limites de Concordância 95%: ({loa_a[0]:.3f}, {loa_a[1]:.3f})")
    print(f"b* -> Viés: {bias_b:.3f} | Limites de Concordância 95%: ({loa_b[0]:.3f}, {loa_b[1]:.3f})")
    print("=======================================================")

# Uso (exemplo para o resultado global)
imprimir_relatorio_avancado(Y_test_geral, prev_geral)
```

```python
from sklearn.model_selection import GroupShuffleSplit
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor, GradientBoostingRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
# =======================================================
# SCRIPT DE ABLAÇÃO AUTOMÁTICA (GRID + ALGORITMOS)
# =======================================================
print("Iniciando Experimentos de Ablação...")
grids_para_testar = [1, 4, 9, 16]
# Envolvemos o GradientBoostingRegressor no MultiOutputRegressor para ele conseguir prever L, a e b juntos!
modelos_para_testar = {
    'Random Forest': lambda: RandomForestRegressor(n_estimators=200, max_depth=15, random_state=42, n_jobs=-1),
    'Extra Trees': lambda: ExtraTreesRegressor(n_estimators=200, max_depth=15, random_state=42, n_jobs=-1),
    'Gradient Boosting': lambda: MultiOutputRegressor(GradientBoostingRegressor(n_estimators=200, max_depth=15, random_state=42))
}
resultados_ablacao = []

# Garante que temos a lista de todas as fotos lidas corretamente
todas_as_fotos = []
for pasta_atual in caminhos_fotos:
    if os.path.exists(pasta_atual):
        for arquivo in os.listdir(pasta_atual):
            if arquivo.lower().endswith(('.jpg', '.png')):
                todas_as_fotos.append((pasta_atual, arquivo))

# LOOP 1: Varia o tamanho do Grid
for grid in grids_para_testar:
    print(f"\n{'='*50}")
    print(f" EXTRAÇÃO DE IMAGENS COM GRID: {grid}")
    print(f"{'='*50}")

    dataset_temp = []

    for pasta_atual, arquivo in tqdm(todas_as_fotos, desc=f"Extraindo Grid {grid}", unit="foto"):
        caminho_completo = os.path.join(pasta_atual, arquivo)
        match = re.search(r'\d+', arquivo)
        if not match: continue

        animal_id = match.group()
        tipo = "pele" if "pele" in arquivo.lower() else "pelo"

        linha = df_colorimetro[df_colorimetro['ANIMAL'] == animal_id]
        if linha.empty: continue

        # CORREÇÃO AQUI: As colunas agora estão no padrão L_pele, a_pele etc.
        if tipo == "pele":
            l, a, b = linha[['L_pele', 'a_pele', 'b_pele']].iloc[0]
        else:
            l, a, b = linha[['L_pelo', 'a_pelo', 'b_pelo']].iloc[0]

        if pd.isna(l): continue

        # Extração
        features_lista = extrair_grid_multicanal(caminho_completo, num_partes=grid)
        if not features_lista: continue

        for f in features_lista:
            dataset_temp.append({
                'ANIMAL': animal_id,
                'R': f['R'], 'G': f['G'], 'B': f['B'],
                'H': f['H'], 'S': f['S'], 'V': f['V'],
                'L_cv2': f['L_cv2'], 'a_cv2': f['a_cv2'], 'b_cv2': f['b_cv2'],
                'L': l, 'a': a, 'b': b,
                'Tipo_Pele': 1 if tipo == "pele" else 0
            })

    df_ml_temp = pd.DataFrame(dataset_temp)
    if df_ml_temp.empty:
        print(f" -> Nenhum dado foi gerado para o Grid {grid}!")
        continue

    df_ia = df_ml_temp.dropna().copy()

    # Engenharia de Features
    df_ia['Razao_RG'] = df_ia['R'] / (df_ia['G'] + 0.000001)
    df_ia['Razao_RB'] = df_ia['R'] / (df_ia['B'] + 0.000001)
    df_ia['Razao_GB'] = df_ia['G'] / (df_ia['B'] + 0.000001)
    df_ia['Dif_L_V'] = df_ia['L_cv2'] - df_ia['V']
    df_ia['Soma_RGB'] = df_ia['R'] + df_ia['G'] + df_ia['B']

    features = ['R', 'G', 'B', 'H', 'S', 'V', 'L_cv2', 'a_cv2', 'b_cv2', 'Razao_RG', 'Razao_RB', 'Razao_GB', 'Dif_L_V', 'Soma_RGB']
    targets = ['L', 'a', 'b']

    # Split
    grupos = df_ia['ANIMAL']
    separador = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
    indice_treino, indice_teste = next(separador.split(df_ia[features], df_ia[targets], groups=grupos))

    df_treino = df_ia.iloc[indice_treino].copy()
    df_teste = df_ia.iloc[indice_teste].copy()

    print(f" -> Grid {grid} gerou {len(df_treino)} amostras de treino e {len(df_teste)} de teste.")

    # LOOP 2: Varia os Algoritmos
    for nome_modelo, funcao_modelo in modelos_para_testar.items():
        print(f" -> Treinando: {nome_modelo}...")

        # Especialista PELE
        treino_pele = df_treino[df_treino['Tipo_Pele'] == 1]
        teste_pele = df_teste[df_teste['Tipo_Pele'] == 1]

        mod_pele = funcao_modelo()
        mod_pele.fit(treino_pele[features], treino_pele[targets])
        prev_pele = pd.DataFrame(mod_pele.predict(teste_pele[features]), columns=['L_pred', 'a_pred', 'b_pred'])
        Y_test_pele = teste_pele[targets].reset_index(drop=True)

        # Especialista PELO
        treino_pelo = df_treino[df_treino['Tipo_Pele'] == 0]
        teste_pelo = df_teste[df_teste['Tipo_Pele'] == 0]

        mod_pelo = funcao_modelo()
        mod_pelo.fit(treino_pelo[features], treino_pelo[targets])
        prev_pelo = pd.DataFrame(mod_pelo.predict(teste_pelo[features]), columns=['L_pred', 'a_pred', 'b_pred'])
        Y_test_pelo = teste_pelo[targets].reset_index(drop=True)

        # Junta os Resultados (Geral)
        Y_test_geral = pd.concat([Y_test_pele, Y_test_pelo]).reset_index(drop=True)
        prev_geral = pd.concat([prev_pele, prev_pelo]).reset_index(drop=True)

        # Métricas globais
        mae_L = mean_absolute_error(Y_test_geral['L'], prev_geral['L_pred'])
        mae_a = mean_absolute_error(Y_test_geral['a'], prev_geral['a_pred'])
        mae_b = mean_absolute_error(Y_test_geral['b'], prev_geral['b_pred'])

        delta_e = np.sqrt((Y_test_geral['L'] - prev_geral['L_pred'])**2 +
                          (Y_test_geral['a'] - prev_geral['a_pred'])**2 +
                          (Y_test_geral['b'] - prev_geral['b_pred'])**2)
        delta_e_mean = np.mean(delta_e)

        r2 = r2_score(Y_test_geral, prev_geral)

        # Guarda na Tabela de Ablação
        resultados_ablacao.append({
            'Algoritmo': nome_modelo,
            'Grid': grid,
            'Amostras (Treino/Teste)': f"{len(df_treino)} / {len(df_teste)}",
            'MAE L*': round(mae_L, 3),
            'MAE a*': round(mae_a, 3),
            'MAE b*': round(mae_b, 3),
            'Delta E': round(delta_e_mean, 3),
            'R2': round(r2, 3)
        })

# =======================================================
# RELATÓRIO FINAL
# =======================================================
df_ablacao = pd.DataFrame(resultados_ablacao)
df_ablacao = df_ablacao.sort_values(by=['Delta E']) # Ordena do melhor pro pior erro

print("\n\n" + "="*80)
print("             TABELA FINAL DE ABLAÇÃO (Ordenado por melhor Delta E)")
print("="*80)
print(df_ablacao.to_string(index=False))

# Salva para o artigo
df_ablacao.to_csv("tabela_de_ablacao.csv", index=False)
print("\n-> Resultados salvos em 'tabela_de_ablacao.csv' com sucesso!")
```
